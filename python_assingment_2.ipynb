```python
# Dictionary (Creation, Modification and Access):
# Create a dictionary named student_marks that maps the names of five
# students to their marks (use scale of from 0 to 100).
student_marks = {
    "chithra": 85,
    "devi": 92,
    "balaji": 78,
    "ashok": 88,
    "shambhu": 95
}
print("student_marks dictionary:", student_marks)
```

```python
# Access and print the mark of a specific student,of your choice
chithra_mark = student_marks["chithra"]
print("Chithra's mark:", chithra_mark)
```

```python
# add a new student"janani" with amark of 80 to the student_marks dictionary.
student_marks["janani"] = 80
print("updated student_marks:",student_marks)
```

```python
# update the mark of any one older student to 82.
student_marks["chithra"] = 82
print("updated student_marks:",student_marks)
```

```python
# use the keys(),values(),and items()methods to print all keys,values, and key-value pairs in the student_marks dictionary.
print("Keys (Names):", list(student_marks.keys()))
print("Values (Marks):", list(student_marks.values()))
print("Items (Key-Value Pairs):", list(student_marks.items()))
```

```python
# set(operations)
# create a set called my_set with following values:['a','e','i','o','u','a','a','i']
# analyse the output and provide explanation for the same.
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print(my_set)
```

```python
# attempt to change the value of my_set[4]='a'.if code throw an error,provide an explanation.
try:
    my_set[4] = 's'
except TypeError as e:
    print(f"Error: {e}")
    print("Explanation: Sets are unordered collections and do not support indexing or item assignment.")
```

```python
# create two sets:
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}


print("set1:", set1)
print("set2:", set2)
```

```python
# compute and print the union and intersection of set1 and set2
union_result = set1.union(set2)
intersection_result = set1.intersection(set2)


print("Union:", union_result)
print("Intersection:", intersection_result)
```

```python

# performance category program.
# prompt user for input.score range should be from 0 to 10.
#find the performance category based on the input score.
#print the performance category.
score = float(input("Enter your score (0 to 10): "))


```

```python
if score < 0 or score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")

elif score > 7:
    print("Performance Category: Above Average")
    print("Excellent performance! Keep up the good work and continue your efforts.")

elif score >= 4:
    print("Performance Category: Average")
    print("Good effort! With more practice and dedication, you can achieve better results.")

else:
    print("Performance Category: Below Average")
    print("You need to improve your performance. Consistent practice will lead to better results.")
```
